# OpenSees Model Analysis 


<details>
<summary><strong>1. Models</strong></summary>

## 1.1 Elastic Elements

* **OpenSees Elastic Beam Column Element**

  * Elastic, uncoupled axial and flexural stiffnesses defined at the element level.
  * **User specifies:**

    * `E` (Young's Modulus)
    * `I` (Moment of Inertia)
    * `A` (Cross-sectional Area)

## 1.2 Inelastic Elements

### 1.2.1 Force-Based Beam-Column Element

Two section types:

<details>
<summary><em>1.2.1.1 Uniaxial Section</em></summary>

* Inelastic, uncoupled axial and flexural stiffnesses defined at the section level.
* **OpenSees command:** `section UniaxialSection ...`
* **User specifies:**

  * Axial stiffness `A`
  * Section moment–curvature characteristics via the `uniaxialMaterial` command

</details>

<details>
<summary><em>1.2.1.2 Fiber Section</em></summary>

* Section is discretized into fibers with independent uniaxial materials.
* Program integrates strains across fibers to compute coupled flexural and axial stiffness/strength.
* **OpenSees command:** `section FiberSection ...`
* **User specifies:**

  * Stress–strain properties for each material via `uniaxialMaterial`
  * Section geometry via series of `patch` and `layer` definitions
* **Example Geometries:**

  * RC Rectangular Section
  * Standard AISC W-section

</details>

* Control node at top floor.
* Lateral-load distribution ∝ mass distribution along height.
* **Analysis type:** Static.

<details>

</details>

<details>
<summary><em>2.1.2 Reversed Cyclic Pushover</em></summary>

* Displacement cycles imposed in positive and negative directions.

</details>

<details>

</details>

<details>

</details>

<details>

<details>
<summary><strong>3. Simulation Process</strong></summary>

Each example script follows this workflow:

1. **Build Model**

   * Define dimensions & DOFs
   * Create nodes (coordinates, constraints, masses)
   * Define elements & connectivity
   * Set up recorders for output

2. **Gravity Analysis**

   * Apply nodal/element loads
   * Set static-analysis parameters (tolerance, load increments)
   * Run `analyze` for gravity
   * Hold gravity loads constant & reset time to zero

3. **Lateral Loading**

   * Define load pattern:

     * Nodal loads for static pushover
     * Ground motions for earthquake
   * Set lateral-analysis parameters (tolerance, increments)

4. **Static Lateral-Load Analysis**

   * Define displacement increments & path
   * Run `analyze` for pushover

5. **Dynamic Lateral-Load Analysis**

   * Specify input motion parameters (scaling, type)
   * Define duration & time increment
   * Define damping
   * Run `analyze` for dynamic response

</details>
